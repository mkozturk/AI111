# Error handling with try-except

We have seen many examples of *runtime errors*. Such errors stop the execution of our programs.

```python
a = 0
print(1/a)
```

Different conditions create different errors.

```python
L = [1,2,3,4]
L[6]
```

These predefined errors are called *exceptions*. Python provides a mechanism ("exception handling") to catch these errors without stopping our program.

```python
a = 0
try:
    print(1/a)
except ZeroDivisionError:
    print("Invalid value. Try again.")
```

One can use `except` without an exception name. Then, it catches any error (the most general error, called `Exception`).

```python
a = 0
try:
    print(1/a)
except:
    print("Invalid value. Try again.")
```

The error message can be captured with the `except ... as ...` syntax and used later:

```python
a = 0
try:
    print(1/a)
except ZeroDivisionError as e:
    print(e)
```

Example: Print the inverses of numbers given by the user.

```python
while True:
    x = input("Enter a number, 'end' to stop:")
    if x=="end":
        break
    try:            
        print(1/int(x))
    except ZeroDivisionError as e:
        print(e)
```

However, this construction does not catch other types of errors, such as a non-numerical entry.

```python
while True:
    x = input("Enter a number, 'end' to stop:")
    if x=="end":
        break
    try:            
        print(1/int(x))
    except ZeroDivisionError as e:
        print(e)
```

To capture both errors, we give them as a tuple of error names:

```python
while True:
    x = input("Enter a number, 'end' to stop:")
    if x=="end":
        break
    try:            
        print(1/int(x))
    except (ZeroDivisionError, ValueError) as e:
        print(e)
```

To capture any exception, use the most general `Exception` value.

```python
while True:
    x = input("Enter a number, 'end' to stop:")
    if x=="end":
        break
    try:            
        print(1/int(x))
    except Exception as e:
        print(e)
```

Best practice is to use the more specific exception rather than the more general one.

A hierarchy of built-in exceptions can be found here: https://docs.python.org/3/library/exceptions.html#exception-hierarchy

If we prefer not to have any action in the event of an error, we can write `pass` in the `except` body:

```python
while True:
    x = input("Enter a number, 'end' to stop:")
    if x=="end":
        break
    try:            
        print(1/int(x))
    except (ZeroDivisionError, ValueError) as e:
        pass
```

The `pass` keyword is used whenever a statement is required by the syntax. It causes no action, and avoids a syntax error.

```python
def incomplete_function(x):
    pass

incomplete_function(1)
```

```python
for i in range(5):
    pass
```

Using the `raise` keyword, we can trigger errors to be caught later in the code.

```python
raise Exception
```

```python
raise ValueError
```

For example, we can raise a `ValueError` if the input is less than zero.

```python
def interest(capital, interest_rate):
    if interest_rate < 0:
        raise ValueError("Interest rate cannot be less than zero.")
    return capital*interest_rate

interest(10000, -0.2)
```

## Assertions

The `assert` keyword, raises an `AssertionError` if the following value is `False.

```python
assert 1<=1
```

```python
assert 1<1
```

This is a very useful tool to verify that the inputs of our functions conform to our expectations.

```python
def inverse(x):
    assert x!=0, "Parameter should not be zero"
    assert type(x) in [int, float], "Parameter should be numeric"
    return 1/x

inverse(2)
```

```python
inverse(0)
```

```python
inverse("asdf")
```
