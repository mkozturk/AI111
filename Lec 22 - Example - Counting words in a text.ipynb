# Example: Counting words in a long piece of text

We will analyze the word frequencies in the The Sherlock Holmes novel *A Study in Scarlet* by Arthur Conan Doyle.

Our goal is to break this text into words and calculate the frequency (count) of every word.

```python
# load the text from the web
file_url = "https://gutenberg.org/cache/epub/244/pg244.txt"
from urllib.request import urlopen
with urlopen(file_url) as response:
    text = response.read().decode("utf-8")
```

Let's look at the first 1000 characters

```python
print(text[:1000])
```

We need to:
* Break the text into words
* Remove nonalphabetic characters (digits, punctuation, spaces, ...)
* Convert to lowercase letters
* Count every word

We use a dictionary, keyed by words, to store the count of words. The count for a particular word is incremented by one at every occurrence of that word.

Our first task is to *preprocess* the text: Generate a list of lowercase words.

Let's separate a small part of the text to serve as testing ground. We can simply use `split()` to break the text at whitespace.

```python
words = text[9200:9400].split()
```

The resulting list of strings has words with some nonalphabetic characters.

```python
words
```

After iterative development, we create the following function that takes one word from the list above, removes nonalphabetic characters in it, and returns the word in lowercase.

```python
def transform(word):
    """Remove digits and punctuation and convert to lowercase."""
    newword = ""
    for c in word: # iterate over characters in the word
        if c.isalpha(): # if the character is alphabetical
            newword += c.lower()  # add the lowercase form of the character to the newword string
    return newword
transform(words[0])
```

Now we can apply this method to the entire text:

```python
words = [transform(w) for w in text.split()]
words[:10]
```

Now that we have a preprocessed list of words, we can start counting them.

We start with an empty dictionary and iterate over the list of words. If the word is already found as a key in the dictionary, we add one to its value. If no such key is found, we set the value to one.

```python
word_count = {} # empty dict
```

```python
for w in words:
    if w in word_count:
        word_count[w] += 1
    else:
        word_count[w] = 1
```

Now, given a word, we can see how many times it occurs in the text:

```python
word_count["holmes"]
```

```python
word_count["the"]
```

The number of unique words in the text is simply the length of this dictionary. Note that this is not the length of the original list of words.

```python
print(f"There are {len(words)} words in the text and {len(word_count)} unique words.")
```

To see the most frequently used words, let's sort by word count.

A dictionary cannot be sorted by itself. So, we sort the output of `word_count.items()` by the second element (values) in reverse order.

```python
sorted_counts = sorted(word_count.items(), key=lambda x:x[1], reverse=True)
sorted_counts[:20] # show the most common 20 words
```
