# Modules

A *module* is a Python script containing definitions of functions and classes you would reuse in your programs.

A module is *imported* with the `import` keyword using the name of the file (without the `.py` extension).

When a module is imported, all contents of the module file is executed.

When the `import` statement is repeated, the file is **not** executed again.

Example:

```
# mymodule.py
myvar = 1234

print("Hello, how are you today?")

def add(x,y):
    return x+y

def multiply(x,y):
    return x*y
```


Import this module

````python
import mymodule
````

The `import` statement executes the module as a source file. That's why we see the output of `print()`.

After importing, we'll be able to access the names in the module using the `mymodule.<name>` notation:

````python
mymodule.myvar
````

````python
mymodule.add(2,3)
````

As usual, we can use the `import ... as ...` form to give a shorter alias to the module.

````python
import mymodule as mm
````

````python
mm.add(2,3)
````

The names in a module can be listed with the `dir()` keyword.

````python
dir(mymodule)
````

An import is executed only once in an interpreter session:

````python
import mymodule
````

Note that the output of `print` is not displayed now, because the module is not executed again.

If you change the source code in the module and want to wor with the new version, you'll have to restart the interpreter (or Jupyter kernel), or use the `reload()` function from the standard library.

````python
import importlib
mym = importlib.reload(mymodule)
````

````python
mym.multiply(4,6)
````
